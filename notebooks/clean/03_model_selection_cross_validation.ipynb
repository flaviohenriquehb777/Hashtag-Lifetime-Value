# 03 – Model Selection & Cross Validation

Nesta etapa, utilizamos a base final limpa (`ltv_base_final.csv`) para rodar os 4 modelos (Dummy, Regressão Linear, Regressão Polinomial e Random Forest) em um pipeline robusto com validação cruzada K-Fold = 5.

```python
import pandas as pd
import sys
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

PROJECT_ROOT = Path.cwd().resolve()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / "config" / "config.yaml").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

from src.models import evaluate_models_on_final_base

# Carregando a base final tratada
df_final = pd.read_csv(PROJECT_ROOT / "data" / "final" / "ltv_base_final.csv")
display(df_final.head())
```

## Resultados dos Modelos (R²)

```python
# Executando a modelagem com KFold = 5 e gerando os pipelines
df_results, summary, split_info = evaluate_models_on_final_base(df_final, random_state=42, n_splits=5)

# Formatando a tabela de resultados: Modelos nas linhas, Folds e Média nas colunas
tabela_final = df_results.T
tabela_final["Média R²"] = summary["Média R²"]
tabela_final["Desvio Padrão"] = summary["Desvio Padrão"]

# Ordenando do maior para o menor R² médio
tabela_final = tabela_final.sort_values(by="Média R²", ascending=False)

print("=== Resultados da Validação Cruzada (KFold=5) ===")
display(tabela_final)
```
